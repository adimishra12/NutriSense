```python
NHANES_RENAME = {
    "SEQN": "person_id",
    "RIAGENDR": "gender",
    "RIDAGEYR": "age_years",

    "DR1TKCAL": "calories_kcal",
    "DR1TPROT": "protein_g",
    "DR1TIRON": "iron_mg",
    "DR1TCALC": "calcium_mg",
    "DR1TVD": "vitamin_d_mcg",
    "DR1TVB12": "vitamin_b12_mcg",
    "DR1TFOLA": "folate_mcg",
    "DR1TVARA": "vitamin_a_mcg",
    "DR1TVC": "vitamin_c_mg",
    "DR1TMAGN": "magnesium_mg",
    "DR1TZINC": "zinc_mg",

    "LBXHGB": "hemoglobin_g_dl",
    "LBXVIDMS": "serum_vitamin_d",
    "LBXFER": "ferritin",
}
```

```python
DEMO_CDC_COLS = [
    "SEQN",
    "RIAGENDR",
    "RIDAGEYR",
]

DIET_CDC_COLS = [
    "SEQN",
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TIRON",
    "DR1TCALC",
    "DR1TVD",
    "DR1TVB12",
    "DR1TFOLA",
    "DR1TVARA",
    "DR1TVC",
    "DR1TMAGN",
    "DR1TZINC",
]

CBC_CDC_COLS = [
    "SEQN",
    "LBXHGB",
]

VID_CDC_COLS = [
    "SEQN",
    "LBXVIDMS",
]

FER_CDC_COLS = [
    "SEQN",
    "LBXFER",
]
```

```python
def read_xpt(path: Path, cdc_cols: list) -> pd.DataFrame:
    frame = pd.read_sas(path, format="xport")

    keep = [c for c in cdc_cols if c in frame.columns]

    missing = set(cdc_cols) - set(keep)

    if missing:
        print(f"Warning: {path.name} missing {missing}")

    frame = frame[keep]

    frame.rename(columns=NHANES_RENAME, inplace=True)

    return frame
```

```python
def evaluate_multilabel(name, y_true, y_pred, label_names):
    """
    Evaluate a multi-label classifier.
    Prints overall metrics and per-label F1 scores.
    """

    accuracy = accuracy_score(
        y_true.values.ravel(),
        y_pred.ravel()
    )

    f1_macro = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1_micro = f1_score(
        y_true,
        y_pred,
        average="micro",
        zero_division=0
    )

    f1_each = f1_score(
        y_true,
        y_pred,
        average=None,
        zero_division=0
    )

    print("\n" + "=" * 70)
    print(f"SCORES — {name}")
    print("=" * 70)

    print(f"Accuracy (overall): {accuracy:.4f} ({accuracy*100:.2f}%)")
    print(f"F1 macro:          {f1_macro:.4f}")
    print(f"F1 micro:          {f1_micro:.4f}")

    print("\nPer-label F1")

    score_df = pd.DataFrame({
        "Label": label_names,
        "F1 Score": np.round(f1_each, 4)
    })

    display(score_df)

    return {
        "Model": name,
        "Accuracy": accuracy,
        "Macro F1": f1_macro,
        "Micro F1": f1_micro,
        "PerLabel": score_df
    }
```

```python
def predict_proba_matrix(model, x_matrix):
    """
    Stack probability(class=1) from each MultiOutput head.
    """

    prob_list = model.predict_proba(x_matrix)

    columns = []

    for probs in prob_list:

        if probs.shape[1] == 1:
            columns.append(np.zeros(len(x_matrix)))
        else:
            columns.append(probs[:, 1])

    return np.column_stack(columns)
```
