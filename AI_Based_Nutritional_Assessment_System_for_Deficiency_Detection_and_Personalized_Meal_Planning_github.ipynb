```python
NHANES_RENAME = {
    "SEQN": "person_id",
    "RIAGENDR": "gender",
    "RIDAGEYR": "age_years",

    "DR1TKCAL": "calories_kcal",
    "DR1TPROT": "protein_g",
    "DR1TIRON": "iron_mg",
    "DR1TCALC": "calcium_mg",
    "DR1TVD": "vitamin_d_mcg",
    "DR1TVB12": "vitamin_b12_mcg",
    "DR1TFOLA": "folate_mcg",
    "DR1TVARA": "vitamin_a_mcg",
    "DR1TVC": "vitamin_c_mg",
    "DR1TMAGN": "magnesium_mg",
    "DR1TZINC": "zinc_mg",

    "LBXHGB": "hemoglobin_g_dl",
    "LBXVIDMS": "serum_vitamin_d",
    "LBXFER": "ferritin",
}
```

```python
DEMO_CDC_COLS = [
    "SEQN",
    "RIAGENDR",
    "RIDAGEYR",
]

DIET_CDC_COLS = [
    "SEQN",
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TIRON",
    "DR1TCALC",
    "DR1TVD",
    "DR1TVB12",
    "DR1TFOLA",
    "DR1TVARA",
    "DR1TVC",
    "DR1TMAGN",
    "DR1TZINC",
]

CBC_CDC_COLS = [
    "SEQN",
    "LBXHGB",
]

VID_CDC_COLS = [
    "SEQN",
    "LBXVIDMS",
]

FER_CDC_COLS = [
    "SEQN",
    "LBXFER",
]
```

```python
def read_xpt(path: Path, cdc_cols: list) -> pd.DataFrame:
    frame = pd.read_sas(path, format="xport")

    keep = [c for c in cdc_cols if c in frame.columns]

    missing = set(cdc_cols) - set(keep)

    if missing:
        print(f"Warning: {path.name} missing {missing}")

    frame = frame[keep]

    frame.rename(columns=NHANES_RENAME, inplace=True)

    return frame
```

```python
demo = read_xpt(RAW / "DEMO_L.XPT", DEMO_CDC_COLS)

diet = read_xpt(RAW / "DR1TOT_L.XPT", DIET_CDC_COLS)

cbc = read_xpt(RAW / "CBC_L.XPT", CBC_CDC_COLS)

vid = read_xpt(RAW / "VID_L.XPT", VID_CDC_COLS)

ferritin_df = read_xpt(RAW / "FERTIN_L.XPT", FER_CDC_COLS)
```

```python
raw_tables = {
    "demographics": demo,
    "diet_nutrients": diet,
    "hemoglobin_lab": cbc,
    "vitamin_d_lab": vid,
    "ferritin_lab": ferritin_df,
}

for name, df in raw_tables.items():
    print("\n" + "="*60)
    print(name)
    print("="*60)
    print(df.shape)
    display(df.head(10))
```

```python

# IMPORT LIBRARIES

import numpy as np
import json

# LABEL RULES

LABEL_RULES = {
    "label_iron": "Lab: low Hb (Male<13 / Female<12) OR ferritin<30; else iron intake <8 mg",
    "label_vitamin_d": "Lab: serum Vitamin D <50 nmol/L; else intake <15 mcg",
    "label_vitamin_b12": "Vitamin B12 intake <2.4 mcg",
    "label_calcium": "Calcium intake <1000 mg",
    "label_folate": "Folate intake <400 mcg",
    "label_vitamin_a": "Vitamin A intake <700 mcg",
    "label_vitamin_c": "Vitamin C intake <75 mg",
    "label_magnesium": "Magnesium intake <320 mg",
    "label_zinc": "Zinc intake <8 mg",
    "label_protein": "Protein intake <50 g"
}

# CREATE DEFICIENCY LABELS


# Iron Deficiency

low_hemoglobin = (
    ((df["gender"] == 1) & (df["hemoglobin_g_dl"] < 13))
    |
    ((df["gender"] == 2) & (df["hemoglobin_g_dl"] < 12))
)

low_ferritin = df["ferritin"] < 30

lab_iron_risk = low_hemoglobin | low_ferritin

has_iron_lab = (
    df["hemoglobin_g_dl"].notna()
    |
    df["ferritin"].notna()
)

low_iron_intake = df["iron_mg"] < 8

df["label_iron"] = np.where(
    has_iron_lab,
    lab_iron_risk.fillna(False),
    low_iron_intake
).astype(int)


# Vitamin D Deficiency

has_vitd_lab = df["serum_vitamin_d"].notna()

lab_vitd_risk = df["serum_vitamin_d"] < 50

low_vitd_intake = df["vitamin_d_mcg"] < 15

df["label_vitamin_d"] = np.where(
    has_vitd_lab,
    lab_vitd_risk,
    low_vitd_intake
).astype(int)


# Intake-based Deficiencies

df["label_vitamin_b12"] = (df["vitamin_b12_mcg"] < 2.4).astype(int)

df["label_calcium"] = (df["calcium_mg"] < 1000).astype(int)

df["label_folate"] = (df["folate_mcg"] < 400).astype(int)

df["label_vitamin_a"] = (df["vitamin_a_mcg"] < 700).astype(int)

df["label_vitamin_c"] = (df["vitamin_c_mg"] < 75).astype(int)

df["label_magnesium"] = (df["magnesium_mg"] < 320).astype(int)

df["label_zinc"] = (df["zinc_mg"] < 8).astype(int)

df["label_protein"] = (df["protein_g"] < 50).astype(int)


# LABEL COLUMN LIST

LABEL_COLS = [
    "label_iron",
    "label_vitamin_d",
    "label_vitamin_b12",
    "label_calcium",
    "label_folate",
    "label_vitamin_a",
    "label_vitamin_c",
    "label_magnesium",
    "label_zinc",
    "label_protein",
]


# REQUIRED FEATURES

FEATURE_NEEDED = [
    "gender",
    "age_years",
    "calories_kcal",
    "protein_g",
    "iron_mg",
    "calcium_mg",
    "vitamin_d_mcg",
    "vitamin_b12_mcg",
    "folate_mcg",
    "vitamin_a_mcg",
    "vitamin_c_mg",
    "magnesium_mg",
    "zinc_mg",
]

df = df.dropna(subset=FEATURE_NEEDED).copy()

print("Rows after labels / diet dropna:", df.shape)

# PREVIEW

preview_cols = [
    "person_id",
    "gender",
    "age_years",
    "iron_mg",
    "hemoglobin_g_dl",
    "ferritin",
    "serum_vitamin_d",
] + LABEL_COLS

preview_cols = [c for c in preview_cols if c in df.columns]

print("\nFirst 10 rows AFTER LABELS:\n")

display(df[preview_cols].head(10))

# LABEL DISTRIBUTION


print("\nLabel Balance (0 vs 1)\n")

for col in LABEL_COLS:

    counts = df[col].value_counts().to_dict()

    positive_rate = df[col].mean() * 100

    print(
        f"{col:<22} -> {counts} | Positive = {positive_rate:.2f}%"
    )


# SAVE LABEL RULES

with open(MODELS / "label_rules.json", "w") as f:
    json.dump(LABEL_RULES, f, indent=4)

print("\nSaved:", MODELS / "label_rules.json")


# SAVE LABELLED DATASET

labelled_path = MODELS / "nhanes_with_labels.parquet"

df.to_parquet(
    labelled_path,
    index=False
)

print("\nSaved labelled dataset:")
print(labelled_path)

# FINAL SUMMARY

print("\nFinal Dataset Shape:", df.shape)

display(df.head(10))
```
